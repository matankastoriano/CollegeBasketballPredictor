```python
df = df[df["POSTSEASON"].notna()].copy()
```

```python
X = df.drop(columns="TEAM")
```

```python
X = pd.get_dummies(X, columns=["CONF"], dtype = int)
```

```python
tourney = {
    "Champions": (6, 6),   # played 6, won 6
    "2ND":       (6, 5),   # played 6, won 5 (lost the final)
    "F4":        (5, 4),   # lost in the national semifinal
    "E8":        (4, 3),
    "S16":       (3, 2),
    "R32":       (2, 1),
    "R64":       (1, 0),
    "R68":       (1, 0)
}
```

```python
X["t_games"] = X["POSTSEASON"].map({k: v[0] for k, v in tourney.items()})
X["t_wins"]  = X["POSTSEASON"].map({k: v[1] for k, v in tourney.items()})
X["t_games"] = X["t_games"].fillna(0)
```

```python
X.head()
```

```python
X["G"] = X["G"] - X["t_games"]
X["W"] = X["W"] - X["t_wins"]
```

```python
X[X["G"] >= 37][[ "YEAR", "POSTSEASON", "G", "W"]]
```

```python
X["champion"] = (X["POSTSEASON"] == "Champions").astype(int)
```

```python
X["champion"].value_counts()
```

```python
X = X.drop(columns=["POSTSEASON", "t_games", "t_wins"])
```

```python
train = X[X["YEAR"] <= 2021]
test  = X[X["YEAR"] >= 2022]
```

```python
train.head()
```

```python
exclude = ["champion"]
feature_cols = [c for c in X.columns if c not in exclude]
print(feature_cols)
```

```python
y = X
y = y.drop(columns = feature_cols)
y.head()
X = X.drop(columns = "champion")
```

```python
X_train, y_train = train[feature_cols], train["champion"]
X_test,  y_test  = test[feature_cols],  test["champion"]
```

```python
clf_xgb = xgb.XGBClassifier(
    objective="binary:logistic",
    eval_metric="aucpr",
    n_estimators=300, max_depth=3, learning_rate=0.05,
    random_state=42,
)
clf_xgb.fit(X_train, y_train,   # deliberately left for you — see below
            verbose=True)
```

```python
df_raw = pd.read_csv("cbb.csv")   # if you don't already have it loaded

results = test[["YEAR", "champion"]].copy()
results["TEAM"] = df_raw.loc[test.index, "TEAM"]
results["prob"] = clf_xgb.predict_proba(X_test)[:, 1]

for year, season in results.groupby("YEAR"):
    ranked = season.sort_values("prob", ascending=False).reset_index(drop=True)
    top_pick = ranked.loc[0, "TEAM"]
    champ_rank = ranked.index[ranked["champion"] == 1][0] + 1  # +1: rank 1 = best
    champ_name = ranked.loc[champ_rank - 1, "TEAM"]
    print(f"{year}: model's #1 = {top_pick:<20} | real champion = {champ_name:<20} ranked #{champ_rank}")
```

```python
results["POSTSEASON"] = df_raw.loc[test.index, "POSTSEASON"]
f4_labels = {"Champions", "2ND", "F4"}
results["made_f4"] = results["POSTSEASON"].isin(f4_labels)

for year, season in results.groupby("YEAR"):
    ranked = season.sort_values("prob", ascending=False)
    model_top4 = set(ranked.head(4)["TEAM"])
    real_f4 = set(season[season["made_f4"]]["TEAM"])
    hits = model_top4 & real_f4          # set intersection
    print(f"{year}: model top 4 = {sorted(model_top4)}")
    print(f"      real F4     = {sorted(real_f4)}")
    print(f"      overlap: {len(hits)}/4 {sorted(hits)}\n")
```
